```python
def read_results_file(search_path="resultsOnetime/Results/"):
    """
    This function read the result file thas is combined from many runs in Results folder and
    Generate the plots
    """
    for file_name in glob.iglob(search_path+'*.xlsx', recursive=True):
#           print("Results" in file_name ,"file_name", file_name)
        if  "~$" not in file_name: #"all_combined" in file_name and
            df = pd.read_excel(file_name, header=0)
            return df
```
